Exercici 1 · Qui rep més enllaços? (2 punts)

a) Calcula el grau d'entrada de cada host i mostra els 10 amb més enllaços entrants (nom, etiqueta i valor).

b) Compara aquest top 10 amb el top 10 del PageRank (ALPHA = 0.91). Quants hosts tenen en comú?

c) Explica per què el grau d'entrada i el PageRank donen (o no donen) rànquings semblants, i què té el PageRank que el grau d'entrada no té.

```python
# Imports — leave as is

import gzip
import csv
import networkx as nx
import matplotlib.pyplot as plt
INPUT_NODES_FILENAME = "data/webspam-uk2007/webspam_uk2007-nodes.csv.gz"
INPUT_EDGES_FILENAME = "data/webspam-uk2007/webspam_uk2007-edges.csv.gz"
```

```python
hosts={}
name2id={}
with gzip.open(INPUT_NODES_FILENAME , "rt", encoding="utf-8") as input_file: #obrim el file amb la informació
    reader = csv.DictReader(input_file, delimiter=',', quotechar='"') #ens ajuda a llegir tot el csv
    for record in reader:
        nodeid=int(record["nodeid"]) #extreiem el nodeid per poder utilitzar-lo com a valor
        hosts[int(nodeid)]={"name":record['hostname'], "label":record['label']} #entrem la info del record dintre del índex nodeid
        name2id[record['hostname']]=int(nodeid) #posem en el name2id el nodeid per poder-lo utilitzar
g = nx.DiGraph()
id2degree = {}
N = len(hosts)
for nodeid in range(N):
    id2degree[nodeid] = 0
with gzip.open(INPUT_EDGES_FILENAME, "rt", encoding="utf-8") as input_file:
    reader = csv.DictReader(input_file, delimiter=',', quotechar='"')
    for record in reader:
        source = int(record["source"])
        destination = int(record["destination"]) #convertim en valor el id del node d'arribada
        id2degree[destination]+=1
top10=sorted(range(N), key=lambda i:id2degree[i], reverse=True)[:10]
for position, nodeid in enumerate(top10, start=1): 
        name = hosts[nodeid]["name"] #de cada parella de hosts agafem la variable nom
        label = hosts[nodeid]["label"] #de cada parella de hosts agafem la variable label
        print(f"{position}.{name}({label}):{id2degree[nodeid]:,}")
llista1=set(top10)        
```

```python
ITERATIONS = 25
ALPHA = 0.91
id2degree = {}
N = len(hosts)
for nodeid in range(N):
    id2degree[nodeid] = 0
with gzip.open(INPUT_EDGES_FILENAME, "rt", encoding="utf-8") as input_file:
    reader = csv.DictReader(input_file, delimiter=',', quotechar='"')
    for record in reader:
        source = int(record["source"])
        destination = int(record["destination"]) #convertim en valor el id del node d'arribada
        id2degree[source]+=1
pagerank = [1.0/N] * N
pagerank_aux = [0.0] * N
for iteration in range(ITERATIONS): #per cada iteració del PageRAnk
    print(f"Iteration {iteration+1} of {ITERATIONS}")
    with gzip.open(INPUT_EDGES_FILENAME, "rt", encoding="utf-8") as input_file:
        reader=csv.DictReader(input_file,delimiter=',',quotechar='"')
        for record in reader:
            source=int(record["source"]) #agafem el node d'entada igual
            destination=int(record["destination"]) #agafem també el de sortida
            pagerank_aux[destination]+=pagerank[source]/id2degree[source] # el que s'afegira al destí quan s'acabi la iteració
    pagerank_nou=[ALPHA*pagerank_aux[i]+(1-ALPHA)*(1/N) for i in range(N)] #formula bàsica del PageRank evitant nodes pou
    total=sum(pagerank_nou)
    pagerank_nou=[x/total for x in pagerank_nou] #normalitzant el nou PageRank i convertint-lo en llista pel futur
    delta=sum(abs(pagerank_nou[i]-pagerank[i]) for i in range(N)) #utilitzar la linia anterior i calcular la nova delta
    print(f"delta={delta:.4f}")
    pagerank=pagerank_nou #actualitzem per la següent iteració
    pagerank_aux=[0.0]*N #retornem tot a 0
hosts_by_score=sorted(enumerate(pagerank), key=lambda x:x[1], reverse=True)
top=hosts_by_score[:10] #agafo els que em diuen
for position, (nodeid,value) in enumerate(top, start=1): 
    name = hosts[nodeid]["name"] #de cada parella de hosts agafem la variable nom
    label = hosts[nodeid]["label"] #de cada parella de hosts agafem la variable label
    print(f"{position}.{name}({label}):{value:.6f}") #printeja la llista
llista2=set(nodeid for nodeid, value in top)
conjunt=len(llista1 & llista2)
print(conjunt)
    
```

```python
#inicialitzem variables
g = nx.DiGraph()

with gzip.open(INPUT_EDGES_FILENAME, "rt", encoding="utf-8") as input_file:
    reader = csv.DictReader(input_file, delimiter=',', quotechar='"')
    for record in reader:
        source = int(record["source"]) #convertim en valor el id del node d'arribada
        destination = int(record["destination"])  #convertim en valor el id del node de sortida
        if hosts[source]["label"] != "unlabeled" and hosts[destination]["label"] != "unlabeled": #condició si hi ha paraules spam i cap dels dos és no definit
            g.add_edge(hosts[source]['name'], hosts[destination]['name']) #introduim al graf
print(f"Nodes del subgraf: {g.number_of_nodes()}")
count=0
count2=0
for u, v in g.edges():
    label_u = hosts[name2id[u]]["label"] #evaluem el label del node d'entrada
    label_v = hosts[name2id[v]]["label"] #evaluem el label del de sortida
    if label_u=="spam":
        count+=1
        if label_v=="spam":
             count2+=1
percentatge_spam=100*count2/count 
percentatge_nonspam=100*(1-(count2/count))
print(f"Spam: {percentatge_spam:.2f}%")
print(f"Nonspam: {percentatge_nonspam:.2f}%")
```

```python
spam_spam = 0        # origen spam    -> destí spam
spam_nonspam = 0     # origen spam    -> destí nonspam
nonspam_spam = 0     # origen nonspam -> destí spam
nonspam_nonspam = 0  # origen nonspam -> destí nonspam

with gzip.open(INPUT_EDGES_FILENAME, "rt", encoding="utf-8") as input_file:
    reader = csv.DictReader(input_file, delimiter=',', quotechar='"')
    for record in reader:
        source = int(record["source"])
        destination = int(record["destination"])
        etiqueta_s = hosts[source]["label"]
        etiqueta_d = hosts[destination]["label"]

        if etiqueta_s == "unlabeled" or etiqueta_d == "unlabeled":
            continue   # ignorem l'aresta

        if etiqueta_s == "spam" and etiqueta_d == "spam":
            spam_spam += 1
        elif etiqueta_s == "spam" and etiqueta_d == "nonspam":
            spam_nonspam += 1
        elif etiqueta_s == "nonspam" and etiqueta_d == "spam":
            nonspam_spam += 1
        else:
            nonspam_nonspam += 1

total_etiquetades = spam_spam + spam_nonspam + nonspam_spam + nonspam_nonspam
sortida_spam = spam_spam + spam_nonspam            # arestes etiquetades que surten d'un spam
sortida_nonspam = nonspam_spam + nonspam_nonspam   # arestes etiquetades que surten d'un nonspam

p_spam_spam = 100 * spam_spam / sortida_spam
p_nonspam_spam = 100 * nonspam_spam / sortida_nonspam

print("Arestes etiquetades en total:", total_etiquetades)
print(f"P(destí = spam | origen = spam):    {p_spam_spam:.2f}%")
print(f"P(destí = spam | origen = nonspam): {p_nonspam_spam:.2f}%")
```
